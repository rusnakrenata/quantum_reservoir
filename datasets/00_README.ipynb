## Recommended Order

1. **NARMA-10** -- confirms your reservoir has sufficient memory capacity
2. **Mackey-Glass tau=17** -- compare NRMSE against Table 1 of Fujii & Nakajima 2020
3. **Lorenz** -- autonomous attractor reconstruction (most visually compelling)
4. **Henon** -- fast ablation studies (discrete map, no ODE solver needed)

## Evaluation Metric

All QRC papers use **Normalised Root Mean Square Error (NRMSE)**:

```
NRMSE = sqrt( mean((y_pred - y_true)^2) / var(y_true) )
```

NRMSE = 0 is perfect. NRMSE = 1 means predicting the mean (trivial baseline).
